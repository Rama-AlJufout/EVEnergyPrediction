## Data Understanding, Dirty Data Creation, and Data Cleaning

```python
import pandas as pd
import numpy as np
```

```python
# ============================================================
# STEP 1: LOAD THE ORIGINAL DATASET
# ============================================================

# We import the original CSV file.
# We keep this dataset unchanged so we can compare it with
# the dirty and cleaned datasets later.

df_original = pd.read_csv("EV_Energy_Consumption.csv")
```

```python
# Display the number of rows and columns.
# This helps us understand the size of the dataset.

print("Original Dataset Shape:")
print("Rows:", df_original.shape[0])
print("Columns:", df_original.shape[1])
```

```python
# Display the first five rows.
# This helps us understand what the data looks like.

print("\nFirst Five Rows:")
display(df_original.head())
```

```python
# Define columns based on their real meaning
categorical_columns = [
    "Driving_Mode",
    "Road_Type",
    "Traffic_Condition",
    "Weather_Condition"
]

identifier_columns = [
    "Vehicle_ID"
]

datetime_columns = [
    "Timestamp"
]

target_column = "Energy_Consumption_kWh"

# Create the classification table
classification = []

for col in df_original.columns:

    if col == target_column:
        feature_type = "Target - Numerical"

    elif col in categorical_columns:
        feature_type = "Categorical"

    elif col in identifier_columns:
        feature_type = "Identifier"

    elif col in datetime_columns:
        feature_type = "Datetime"

    elif pd.api.types.is_numeric_dtype(df_original[col]):
        feature_type = "Numerical"

    else:
        feature_type = "Needs Review"

    classification.append({
        "Column": col,
        "Data Type": str(df_original[col].dtype),
        "Feature Type": feature_type,
        "Unique Values": df_original[col].nunique()
    })

# Display the result
classification_df = pd.DataFrame(classification)

print(classification_df.to_string(index=False))
```

```python
# ============================================================
# STEP 2: CHECK THE ORIGINAL DATA QUALITY
# ============================================================

# Check missing values in every column.
# Missing values are empty cells that may affect analysis
# and machine learning.

print("\nMissing Values in Original Dataset:")
display(df_original.isnull().sum())
```

```python
# Count exact duplicate rows.
# A duplicate row repeats the same values in every column.

print("\nDuplicate Rows in Original Dataset:")
print(df_original.duplicated().sum())
```

```python
# Display summary statistics for numerical columns.
# These statistics help us understand ranges, averages,
# and possible unusual values.

print("\nOriginal Numerical Statistics:")
display(df_original.describe().T)

# At this point, we have inspected the original dataset
# without changing any of its values.
```

```python
# ============================================================
# STEP 3: CREATE A DIRTY COPY
# ============================================================

# We create a separate copy of the original dataset.
# All intentional problems will be added to this copy.
# The original dataset remains unchanged.

df_dirty = df_original.copy()
```

```python
# We use a fixed random seed so the same rows are selected
# whenever we run this code again.

np.random.seed(42)
```

```python
# ------------------------------------------------------------
# 3A: INTRODUCE MISSING VALUES
# ------------------------------------------------------------

# We deliberately introduce missing values in selected columns.
# This allows us to demonstrate how missing values are detected
# and handled during the cleaning process.

columns_with_missing_values = [
    "Speed_kmh",
    "Battery_State_%",
    "Temperature_C",
    "Tire_Pressure_psi",
    "Energy_Consumption_kWh"
]

for column in columns_with_missing_values:

    # Check that the column exists before changing it.
    if column in df_dirty.columns:

        # Select approximately 20% of the rows.
        number_of_missing_values = max(
            1, int(len(df_dirty) * 0.20)
        )

        selected_rows = np.random.choice(
            df_dirty.index,
            size=number_of_missing_values,
            replace=False
        )

        # Replace the selected values with NaN.
        df_dirty.loc[selected_rows, column] = np.nan
```

```python
# ------------------------------------------------------------
# 3B: INTRODUCE DUPLICATE ROWS
# ------------------------------------------------------------

# We add 20 existing rows to the end of the dataset.
# These rows become exact duplicates.
# This gives us duplicate records to identify and remove.

number_of_duplicates = min(25, len(df_dirty))

duplicate_rows = df_dirty.sample(
    n=number_of_duplicates,
    random_state=42
)

df_dirty = pd.concat(
    [df_dirty, duplicate_rows],
    ignore_index=True
)
```

```python
# ------------------------------------------------------------
# 3C: INTRODUCE EXTREME VALUES
# ------------------------------------------------------------

# We deliberately insert unusual values into selected columns.
# These values are examples for practicing outlier investigation.
# They are not values discovered in the original dataset.

artificial_outliers = {
    "Speed_kmh": 500,
    "Battery_State_%": 150,
    "Battery_Voltage_V": 1000,
    "Temperature_C": 200,
    "Tire_Pressure_psi": 100,
    "Energy_Consumption_kWh": 100
}

for column, unusual_value in artificial_outliers.items():

    if column in df_dirty.columns and len(df_dirty) > 0:

        # Select one row and replace its value with an extreme value.
        selected_row = np.random.choice(df_dirty.index)

        df_dirty.loc[selected_row, column] = unusual_value
```

```python
# Save the dirty dataset separately.
# This lets us keep the original file unchanged.

df_dirty.to_csv(
    "EV_Energy_Consumption_Dirty.csv",
    index=False
)

print("\nDirty Dataset Created Successfully!")
print("Rows:", df_dirty.shape[0])
print("Columns:", df_dirty.shape[1])
```

```python
# ============================================================
# STEP 4: ANALYZE MISSING VALUES
# ============================================================

# Count missing values for every column in the dirty dataset.

missing_report = pd.DataFrame({
    "Missing_Count": df_dirty.isnull().sum(),
    "Duplicte_Count": df_dirty.duplicated().sum(),
})

print("\nMissing Values Report:")
display(missing_report)
print(df_dirty.shape)
```

```python
# Create a copy for the cleaning process.
# We do not clean df_dirty directly because we want to keep
# the dirty dataset available for comparison.

df_clean = df_dirty.copy()
```

```python
# ------------------------------------------------------------
# 4A: HANDLE MISSING TARGET VALUES
# ------------------------------------------------------------

# Energy_Consumption_kWh is our target variable.
# It is the value that the machine learning model will predict.
# We remove rows with a missing target because the correct
# target value is unknown for those records.

target_column = "Energy_Consumption_kWh"

if target_column in df_clean.columns:

    rows_before = len(df_clean)

    df_clean = df_clean.dropna(
        subset=[target_column]
    )

    rows_removed = rows_before - len(df_clean)

    print("\nRows removed due to missing target values:")
    print(rows_removed)
    print("Rows remaining:", len(df_clean))
    print()
```

```python
# ------------------------------------------------------------
# 4B: HANDLE MISSING NUMERICAL VALUES
# ------------------------------------------------------------

# We fill missing numerical feature values with their median.
# The median is the middle value and is less affected by
# extreme values than the mean.

numerical_columns = df_clean.select_dtypes(
    include=np.number
).columns

for column in numerical_columns:

    # Do not fill missing target values.
    if column == target_column:
        continue

    if df_clean[column].isnull().any():

        median_value = df_clean[column].median()

        df_clean[column] = df_clean[column].fillna(
            median_value
        )

        print(
            "Missing values filled in:",
            column,
            "| Method: Median"
        )
```

```python
# ============================================================
# STEP 5: FIND AND REMOVE DUPLICATES
# ============================================================

# Count exact duplicate rows before removal.

duplicates_before = df_clean.duplicated().sum()

print("\nDuplicate Rows Before Cleaning:")
print(duplicates_before)
```

```python
# Remove exact duplicate rows.
# This avoids counting identical records more than once.

df_clean = df_clean.drop_duplicates().copy()
print(df_clean.shape)
```

```python
# Count duplicates again to confirm the result.

duplicates_after = df_clean.duplicated().sum()

print("\nDuplicate Rows After Cleaning:")
print(duplicates_after)
```

```python
# ============================================================
# STEP 6: INVESTIGATE OUTLIERS USING THE IQR METHOD
# ============================================================

# IQR is a statistical method used to flag unusually low
# or high numerical values.
#
# Q1 = 25th percentile
# Q3 = 75th percentile
# IQR = Q3 - Q1
#
# Lower Bound = Q1 - 1.5 * IQR
# Upper Bound = Q3 + 1.5 * IQR
#
# Values outside these limits are flagged for investigation.
# They are not automatically considered errors.
print("\nOutlier Investigation:")

numerical_columns = df_clean.select_dtypes(
    include=np.number
).columns

outlier_report = []

for column in numerical_columns:

    values = df_clean[column].dropna()

    if values.empty:
        continue

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_mask = (
        (df_clean[column] < lower_bound) |
        (df_clean[column] > upper_bound)
    )

    outlier_count = int(outlier_mask.sum())

    outlier_report.append({
        "Column": column,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": outlier_count
    })

outlier_report = pd.DataFrame(outlier_report)

display(outlier_report)
# Important decision:
# We do not delete every IQR outlier.
# A value may be extreme but still represent a real observation.
# We investigate values using the column meaning and valid ranges.
```

```python
# ============================================================
# STEP 7: CHECK CLEARLY INVALID VALUES
# ============================================================

# These are example validity limits based on the meaning
# and units of selected columns.
# They should be checked against the dataset documentation.
# They are not universal rules for every electric vehicle.

valid_ranges = {
    "Speed_kmh": (0, 200),
    "Battery_State_%": (0, 100),
    "Battery_Voltage_V": (0, 600),
    "Temperature_C": (-50, 60),
    "Humidity_%": (0, 100),
    "Wind_Speed_ms": (0, 75),
    "Tire_Pressure_psi": (10, 60)
}

for column, limits in valid_ranges.items():

    if column not in df_clean.columns:
        continue

    minimum, maximum = limits

    # Identify values outside the configured range.
    invalid_mask = (
        (df_clean[column] < minimum) |
        (df_clean[column] > maximum)
    )

    invalid_count = int(invalid_mask.sum())

    print(
        f"\nInvalid values in {column}:",
        invalid_count
    )

    if invalid_count > 0:

        # Replace invalid feature values with NaN.
        # We then fill them using the median of the valid values.
        df_clean.loc[invalid_mask, column] = np.nan

        median_value = df_clean[column].median()

        if pd.notna(median_value):

            df_clean[column] = df_clean[column].fillna(
                median_value
            )
```

```python
# Check the target separately.
# We remove clearly invalid negative energy consumption values.
# Zero is allowed here because its meaning should be verified
# against the dataset documentation and use case.

if target_column in df_clean.columns:

    invalid_target_mask = df_clean[target_column] < 0

    invalid_target_count = int(
        invalid_target_mask.sum()
    )

    print("\nInvalid target values:", invalid_target_count)

    df_clean = df_clean.loc[
        ~invalid_target_mask
    ].copy()
```

```python
# ============================================================
# STEP 8: VERIFY THE CLEANED DATASET
# ============================================================

# Check whether any missing values remain.

print("\nRemaining Missing Values:")
print(df_clean.isnull().sum().sum())

# Check whether any exact duplicate rows remain.

print("\nRemaining Duplicate Rows:")
print(df_clean.duplicated().sum())

# Display the final dataset size.

print("\nCleaned Dataset Shape:")
print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])

# Display the first five cleaned records.

print("\nFirst Five Rows of Cleaned Dataset:")
display(df_clean.head())

# Display summary statistics after cleaning.

print("\nStatistics After Cleaning:")
display(df_clean.describe().T)
```

```python
# ============================================================
# STEP 9: COMPARE THE DIRTY AND CLEANED DATASETS
# ============================================================

# This comparison shows the effect of our cleaning decisions.

comparison = pd.DataFrame({
    "Dirty Dataset": [
        df_dirty.shape[0],
        df_dirty.shape[1],
        int(df_dirty.isnull().sum().sum()),
        int(df_dirty.duplicated().sum())
    ],
    "Cleaned Dataset": [
        df_clean.shape[0],
        df_clean.shape[1],
        int(df_clean.isnull().sum().sum()),
        int(df_clean.duplicated().sum())
    ]
}, index=[
    "Number of Rows",
    "Number of Columns",
    "Missing Cells",
    "Duplicate Rows"
])

print("\nBefore and After Cleaning:")
display(comparison)

# Compare numerical statistics before and after cleaning.
common_numerical_columns = [
    column
    for column in df_dirty.select_dtypes(include=np.number).columns
    if column in df_clean.columns
]

print("\nDirty Dataset Statistics:")
display(df_dirty[common_numerical_columns].describe().T)

print("\nCleaned Dataset Statistics:")
display(df_clean[common_numerical_columns].describe().T)
```

```python
# ============================================================
# STEP 10: SAVE THE CLEANED DATASET
# ============================================================

# Save the final cleaned data to a new CSV file.
# The original and dirty files remain separate.

df_clean.to_csv(
    "EV_Energy_Consumption_Cleaned.csv",
    index=False
)

print("\nCleaning Process Completed Successfully!")
print("Dirty file: EV_Energy_Consumption_Dirty.csv")
print("Cleaned file: EV_Energy_Consumption_Cleaned.csv")
```
